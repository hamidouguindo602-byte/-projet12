```python
def hospitalisation(data):
  if data['Patient addmited to regular ward (1=yes, 0=no)']==1:
    return 'surveillance'
  elif data['Patient addmited to semi-intensive unit (1=yes, 0=no)']==1:
    return 'soins semi-intensives'
  elif data['Patient addmited to intensive care unit (1=yes, 0=no)']==1:
    return 'soins intensifs'
  else:
    return 'inconnu'
```

PRE-TRAITEMENT



```python
data.shape
```

```python
colonnes=['SARS-Cov-2 exam result','Patient age quantile']
```

```python
dat=data[list(boot_set.columns)+list(viral_set.columns)+colonnes]
```

```python
dat.drop(['Influenza A, rapid test','Influenza B, rapid test'],axis=1,inplace=True)
```

```python
dat.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test1=train_test_split(dat,test_size=0.20,random_state=0)
x_train['SARS-Cov-2 exam result'].value_counts()
```

```python
x_test1['SARS-Cov-2 exam result'].value_counts()
```

```python
for col in viral_set.columns:
  print(f'{col:-<50}{viral_set[col].unique()}')
```

```python
encod={'positive':1,'negative':0,'detected':1,'not_detected':0}
```

```python
def feature_engenering(data):
  data['est_malade']=data[viral_set.columns].sum(axis=1)>=1
  data.drop(viral_set.columns,axis=1,inplace=True)
  return data
```

```python
def Encodage(data):
  for col in data.select_dtypes('object').columns:
      data[col]=data[col].map(encod)
  return data
```

```python
def impulsion(data):
  #data['is na']=(data['Hemoglobin']=='nan') | (data['Parainfluenza 4']=='nan')
  return data.dropna(axis=0)
```

```python
def pre_processing(data):
  data=Encodage(data)
  #data=feature_engenering(data)
  data=impulsion(data)
  x=data.drop('SARS-Cov-2 exam result',axis=1)
  y=data['SARS-Cov-2 exam result']
  return x,y
```

```python
x,y=pre_processing(x_train)

```

```python
x_test,y_test=pre_processing(x_test1)
```

```python
from sklearn.metrics import f1_score,classification_report,confusion_matrix
from sklearn.model_selection import learning_curve
from sklearn.pipeline import make_pipeline
from sklearn.feature_selection import SelectKBest,f_classif
from sklearn.preprocessing import PolynomialFeatures
#from slearn.decomposition import PCA
```

```python
model=make_pipeline(PolynomialFeatures(2),SelectKBest(f_classif,k=9),RandomForestClassifier(random_state=0))
model.fit(x,y)
```

```python
def Evaluation(model):
  ypre=model.predict(x_test)
  print(confusion_matrix(y_test,ypre))
  print(classification_report(y_test,ypre))
  n,train_score,val_score=learning_curve(model,x,y,train_sizes=np.linspace(0.1,1,10),scoring='f1')
  plt.figure()
  plt.plot(n,train_score.mean(axis=1),label='train')
  plt.plot(n,val_score.mean(axis=1),label='val')
  plt.legend()
```

```python
Evaluation(model)
```
