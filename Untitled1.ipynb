```python
data=data[data.columns[data.isna().sum()/data.shape[0]<0.9]]
data.drop('Patient ID',axis=1,inplace=True)
```

HISTOGRAMME

```python
for col in data.select_dtypes('float64'):
  plt.figure()
  sns.displot(data[col],kde=True)
```

```python
plt.figure()
sns.displot(data['Patient age quantile'],kde=True)
```

```python
for col in data.select_dtypes('object'):
  print(f'{col:-<40}{data[col].unique()}')
```

```python
for col in data.select_dtypes('object'):
  plt.figure()
  data[col].value_counts().plot.pie()
```

```python
negative_set=data[data['SARS-Cov-2 exam result']=='negative']
positive_set=data[data['SARS-Cov-2 exam result']=='positive']
boot_set=data[data.columns[data.isna().sum()/data.shape[0]>0.88]]
viral_set=data[data.columns[(data.isna().sum()/data.shape[0]<0.88) & (data.isna().sum()/data.shape[0]>0.75)]]
```

```python
for col in boot_set.columns:
  plt.figure(figsize=(10,19))
  sns.histplot(negative_set[col],kde=True,label='negative',stat='density')
  sns.histplot(positive_set[col],kde=True,label='positive',stat='density')
  plt.legend()
```
